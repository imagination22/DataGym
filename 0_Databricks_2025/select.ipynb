## ✅ **What Are Data Retrieval Statements?**

These are SQL commands used to **fetch data from tables or views** in Databricks. They allow you to query rows based on conditions, sort results, and even inspect execution plans.

***

### **Core Statements**

1.  **`SELECT`**
    *   Standard SQL query to retrieve rows from one or more tables.
    *   Syntax:
        ```sql
        SELECT column1, column2
        FROM table_name
        WHERE condition
        ORDER BY column1;
        ```
    *   Supports clauses like `WHERE`, `GROUP BY`, `ORDER BY`, `LIMIT`.

2.  **`VALUES`**
    *   Creates an **inline temporary table** with literal values.
    *   Example:
        ```sql
        VALUES (1, 'Alice'), (2, 'Bob');
        ```

3.  **`Query`**
    *   A general term for any SQL statement that retrieves data.
    *   Can include **subqueries**, **CTEs**, and **set operators** (`UNION`, `INTERSECT`, `EXCEPT`).

***

### **Databricks-Specific Feature: SQL Pipeline Syntax**

*   Instead of writing one big SQL statement, you can **chain operations** using `|>` (pipe operator).
*   Example:
    ```sql
    FROM sales
    |> WHERE amount > 100
    |> SELECT customer_id, amount
    |> ORDER BY amount DESC;
    ```
*   Benefits:
    *   Easier to read.
    *   Modular for complex transformations.

***

### **Inspecting Query Plans: `EXPLAIN`**

*   Use `EXPLAIN` to see **how Databricks will execute your query**:
    *   **Logical Plan**: High-level steps.
    *   **Physical Plan**: Actual execution strategy.
*   Example:
    ```sql
    EXPLAIN SELECT * FROM sales WHERE amount > 100;
    ```
*   Helps with **performance tuning** and understanding query optimization.

***

### ✅ **Why This Matters**

*   Knowing these statements helps you:
    *   Retrieve data efficiently.
    *   Debug and optimize queries.
    *   Use Databricks features like **SQL pipelines** for better readability.

***
