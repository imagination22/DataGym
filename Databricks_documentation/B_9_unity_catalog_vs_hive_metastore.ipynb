Here’s a **detailed explanation of database objects in the legacy Hive metastore and how they differ from Unity Catalog**, with practical examples:

***

## ✅ **What is the Hive Metastore in Databricks?**

*   The **Hive metastore** is the legacy metadata store for tables, views, and schemas in Databricks.
*   In Unity Catalog-enabled workspaces, Hive metastore objects appear under the **`hive_metastore` catalog**.
*   Example query:

```sql
SELECT * FROM hive_metastore.sales.orders;
```

***

## ✅ **Key Differences Between Hive Metastore and Unity Catalog**

1.  **Governance**
    *   Hive metastore uses **table ACLs** (legacy) for access control.
    *   Unity Catalog provides **fine-grained governance** with centralized policies.
    *   Credential passthrough in Hive is deprecated.

2.  **Hierarchy**
    *   Hive metastore: **Schema → Table/View**
    *   Unity Catalog: **Catalog → Schema → Table/View**

3.  **Storage Management**
    *   Hive schemas can specify a **LOCATION** (cloud URI or DBFS path).
    *   Default Hive schema location: `/user/hive/warehouse/<schema-name>` (on DBFS root).
    *   Unity Catalog uses **managed storage locations** with better isolation.

***

## ✅ **Schemas in Hive Metastore**

*   Created using SQL (not Catalog Explorer).
*   Naming: Only **alphanumeric ASCII + underscores**.
*   Example:

```sql
CREATE SCHEMA sales LOCATION 'abfss://container@storageaccount.dfs.core.windows.net/sales_data/';
```

**Important behaviors:**

*   Dropping a schema deletes **all files recursively** in its location.
*   Avoid overlapping schema locations or sharing paths.

***

## ✅ **Managed Tables in Hive Metastore**

*   Stored in the schema’s location.
*   Can use **Delta Lake** or other formats.
*   Dropping a managed table deletes data **immediately** (no recovery like Unity Catalog).
*   Example:

```sql
CREATE TABLE hive_metastore.sales.orders (
    order_id STRING,
    amount DOUBLE
) USING DELTA;
```

***

## ✅ **External Tables in Hive Metastore**

*   Point to existing data in cloud storage.
*   Data is **not deleted** when the table is dropped.
*   Example:

```sql
CREATE TABLE hive_metastore.sales.external_orders (
    order_id STRING,
    amount DOUBLE
)
USING DELTA
LOCATION 'abfss://container@storageaccount.dfs.core.windows.net/external_orders/';
```

***

## ✅ **Views in Hive Metastore**

*   Can reference **any data source**, including raw paths.
*   Risk: Views can bypass ACLs if they use direct URIs.
*   Example:

```sql
CREATE VIEW hive_metastore.sales.high_value_orders AS
SELECT * FROM hive_metastore.sales.orders WHERE amount > 1000;
```

***

## ✅ **Legacy Hive Features**

*   HiveQL support (e.g., `LOAD DATA`, `INSERT OVERWRITE DIRECTORY`).
*   Hive UDFs and compatibility remain but are **deprecated**.

***

## ✅ **Best Practices**

*   **Do not use DBFS root** for production data.
*   Avoid overlapping schema or table locations.
*   Migrate Hive metastore objects to **Unity Catalog** for:
    *   Better governance
    *   Recovery options
    *   Performance optimizations
*   Example migration command:

```sql
ALTER TABLE hive_metastore.sales.orders
SET TBLPROPERTIES ('upgraded_to_unity_catalog' = 'true');
```
