## 📊 Table Schema

```markdown
+---------------+---------+
| Column Name   | Type    |
+---------------+---------+
| username      | varchar |
| activity      | varchar |
| startDate     | Date    |
| endDate       | Date    |
+---------------+---------+
```

- This table does **not** contain a primary key.  
- It contains information about the activity performed by each user in a period of time.  
- A person with `username` performed an `activity` from `startDate` to `endDate`.  

---

## ❓ Question

Write an SQL query to show the **second most recent activity** of each user.  

- If the user only has **one activity**, return that one.  
- A user can't perform more than one activity at the same time.  
- Return the result table in any order.  

---

## 📌 Example

### Input: `UserActivity` table
```markdown
+------------+--------------+-------------+-------------+
| username   | activity     | startDate   | endDate     |
+------------+--------------+-------------+-------------+
| Alice      | Travel       | 2020-02-12  | 2020-02-20  |
| Alice      | Dancing      | 2020-02-21  | 2020-02-23  |
| Alice      | Travel       | 2020-02-24  | 2020-02-28  |
| Bob        | Travel       | 2020-02-11  | 2020-02-18  |
+------------+--------------+-------------+-------------+
```

### Expected Output: Result table
```markdown
+------------+--------------+-------------+-------------+
| username   | activity     | startDate   | endDate     |
+------------+--------------+-------------+-------------+
| Alice      | Dancing      | 2020-02-21  | 2020-02-23  |
| Bob        | Travel       | 2020-02-11  | 2020-02-18  |
+------------+--------------+-------------+-------------+
```

---

Would you like me to also format this into a **ready-to-use SQL practice question sheet** (like a mini assignment template) so you can reuse it easily?

````python
%sql
CREATE TABLE UserActivity (
    username   VARCHAR(50),
    activity   VARCHAR(50),
    startDate  DATE,
    endDate    DATE
);
INSERT INTO UserActivity (username, activity, startDate, endDate)
VALUES
('Alice', 'Travel',  '2020-02-12', '2020-02-20'),
('Alice', 'Dancing', '2020-02-21', '2020-02-23'),
('Alice', 'Travel',  '2020-02-24', '2020-02-28'),
('Bob',   'Travel',  '2020-02-11', '2020-02-18');
````

````python
%sql
with cte as (

Select dense_rank()over(partition by username order by startDate , endDate asc)as rn 
, count(*)over(partition by username)as cnt
,*
from UserActivity

)
select username , activity  , startDate , endDate from cte 
where case 
  when cnt = 1 then rn =1
  else rn =2 
  end
  ;
````

````python
%sql
with cte as (

  select distinct * from UserActivity
)
,
 cte2 as (

Select rank()over(partition by username order by startDate , endDate asc)as rn 
, count(*)over(partition by username)as cnt
,*
from cte 

)
select username , activity  , startDate , endDate from cte2 
where case 
  when cnt = 1 then rn =1
  else rn =2 
  end
  ;
````
