Here’s a **simplified version of Dashboards in Databricks AI/BI** so you can **remember easily**:

***

### ✅ **What is a Dashboard?**

A **dashboard** is a **visual report** that shows data in charts, tables, and graphs so you can **understand and share insights quickly**.

***

### ✅ **Why use AI/BI Dashboards?**

*   **AI-assisted authoring** → Helps you build dashboards faster.
*   **Better visuals** → Enhanced charts and widgets.
*   **Easy sharing** → Share with anyone in your Databricks account (even if they don’t have workspace access).

***

### ✅ **Important Update**

*   Old dashboards = **Legacy dashboards** (support ends **Jan 12, 2026**).
*   New dashboards = **AI/BI dashboards** (recommended).
*   You can **migrate old dashboards** using a tool or API.

***

### ✅ **Main Components**

1.  **Data Tab**
    *   Define datasets (queries or tables) for your dashboard.
    *   Example: SELECT \* FROM sales\_data WHERE region = 'APAC'.

2.  **Canvas Tab**
    *   Build multi-page reports.
    *   Add widgets like charts, filters, text, and images.

***

### ✅ **Key Actions**

*   **Create**: Click **Create** on the dashboard page.
*   **Organize**: Store dashboards in folders in your workspace.
*   **Delete**: Move to trash (auto-deletes after 30 days).
*   **Draft vs Published**:
    *   Draft = Editable version.
    *   Published = Shared version (doesn’t change until republished).

***

### ✅ **Sharing & Permissions**

*   **Publish** dashboard → Choose how viewers access data:
    *   **Share data permission** (default): Viewers use publisher’s credentials.
    *   **Individual data permission**: Viewers use their own credentials.
*   **Notify viewers** → Optional email alerts when publishing.

***

### ✅ **Extra Features**

*   Download as **PDF**.
*   Use **REST API** for automation.
*   Add **trusted datasets** from Unity Catalog for governance.

***

### ✅ **Quick Memory Tip**

Think of **Dashboards** as:

*   **Data + Canvas + Share**  
    → Data defines what you show, Canvas designs how you show, Share lets others see it.

Here’s a **simplified version of “Add pages, visualizations, text, and filters to the canvas”** so you can **remember easily**:

***

### ✅ **What is the Canvas?**

The **Canvas tab** is where you **design your dashboard**.  
Think of it like a **blank page** where you add:

*   **Pages** → Organize content into sections.
*   **Visualizations** → Charts and graphs.
*   **Text** → Notes, titles, or images.
*   **Filters** → Let users drill down into data.

***

### ✅ **Pages**

*   Start with **one page** (Untitled).
*   **Rename**: Double-click the title.
*   **Add**: Click **+** next to the page title.
*   **Delete**: Use the **kebab menu → Delete**.
*   **Clone**: Copy a page with all widgets (kebab menu → Clone).

***

### ✅ **Visualizations**

*   Types: Bar, Line, Pie, Heatmap, Table, Scatter, etc.
*   **Create**:
    *   Use **AI Assistant** → Describe chart in plain language.
    *   Or use **Config Panel** → Pick dataset, X-axis, Y-axis, colors.
*   **Forecasting**: Add **AI Forecast** to line charts for future trends.

***

### ✅ **Text Widgets**

*   Add **formatted text**, **links**, or **images**.
*   Supports **Markdown** for styling.
*   Example: Add a company logo or a note explaining the chart.

***

### ✅ **Filters**

*   Apply at **global**, **page**, or **widget level**.
*   Helps users narrow down data (like slicers in Excel).
*   **Cross-filtering**: Click a chart element → Other charts update automatically.
*   **Drill-through**: Click a data point → Jump to a detailed page filtered by that selection.

***

### ✅ **Extra Features**

*   **Copy & Paste Widgets** → Move between pages.
*   **Keyboard Shortcuts**:
    *   Ctrl+C / Ctrl+V → Copy/Paste widgets.
    *   Ctrl+Z → Undo.
*   **Download**:
    *   Charts → PNG.
    *   Data → CSV, Excel.
*   **Customize Look**: Colors, fonts, themes.
*   **Tags**: Organize dashboards.
*   **Enable Genie**: Adds conversational AI to your dashboard.

***

### ✅ **Quick Memory Tip**

Think of **Canvas** as:
**Pages + Charts + Text + Filters → Organized, Interactive Dashboard**

Here’s a **detailed but simplified explanation of “Use Filters” in Databricks dashboards** so you can understand and remember easily:

***

### ✅ **What Are Filters?**

Filters help you **narrow down data** in your dashboard, just like slicers in Excel or Power BI.  
They make dashboards interactive so viewers can **focus on specific data segments**.

***

### ✅ **Types of Filters**

1.  **Global Filters**
    *   Apply to the **entire dashboard**.
    *   Example: Filter by **Region = APAC** → All pages update.

2.  **Page-Level Filters**
    *   Apply only to **one page**.
    *   Example: Filter by **Month = January** on the “Sales” page.

3.  **Widget-Level Filters**
    *   Apply to **one chart or table**.
    *   Example: Filter a bar chart to show **Top 10 Products**.

***

### ✅ **How Filters Work**

*   Filters use **dataset columns** or **SQL query parameters**.
*   Example: If your dataset has `region`, `month`, `product`, you can create filters for these fields.

***

### ✅ **Copy & Paste Filters**

*   You can **move filters** between:
    *   Canvas → Global filter panel.
    *   Use **Ctrl+C / Ctrl+V** (or Cmd+C / Cmd+V on Mac).
*   **Note**: Works only within the same dashboard.

***

### ✅ **Cross-Filtering**

*   Lets you **click on a chart element** to filter other charts automatically.
*   Example:
    *   Click **“2024”** in a bar chart → All charts update to show 2024 data.
*   Supported chart types:
    *   Bar, Box plot, Heatmap, Histogram, Pie, Scatter, Point map.

***

### ✅ **Drill-Through**

*   Lets you **jump to a detailed page** filtered by your selection.
*   Example:
    *   On “Overview” page, click **January** in a bar chart → Opens “Details” page showing January data.
*   Supported chart types: Same as cross-filtering.
*   **Tip**: Works best when date fields use transformations like DAILY.

***

### ✅ **Best Practices**

*   **Scope your SQL carefully** → Avoid exposing sensitive data.
*   Use **explicit column selection** and **WHERE clauses**.
*   Combine filters with **cross-filtering** for deeper analysis.

***

### ✅ **Quick Memory Tip**

Think of filters as **three levels**:
**Global → Page → Widget**, plus **Cross-filtering** and **Drill-through** for interactive exploration.

***

### ✅ **Requirements**

*   Logged into **Azure Databricks workspace**.
*   Have **Databricks SQL entitlement**.
*   Access to at least one **SQL warehouse**.

***

### ✅ **Steps to Create a Dashboard**

#### **Step 1: Create Dashboard**

*   Click **New → Dashboard**.
*   Dashboard is saved in your **Workspace/Users/<username>** folder.
*   You can also create from **Dashboards listing page**.

***

#### **Step 2: Define Datasets**

*   Go to **Data tab**.
*   Click **Create from SQL**.
*   Write or paste your SQL query → Click **Run**.
*   Rename dataset (e.g., **Taxicab data**).
*   **Tip**: You can also select a **Unity Catalog table/view** instead of writing SQL.

***

#### **Step 3: Add Visualization**

*   Go to **Canvas tab**.
*   Click **Add Visualization** → Place it on the canvas.

***

#### **Step 4: Configure Visualization**

*   Select **Dataset**.
*   Choose **Chart type** (e.g., Bar).
*   Configure:
    *   **X-axis** → tpep\_dropoff\_datetime (Transform: HOURLY).
    *   **Y-axis** → fare\_amount (Transform: AVG).
*   Optional: Use **Databricks Assistant** → Type:
        Bar chart of average fare amount over hourly dropoff time

***

#### **Step 5: Clone & Modify Visualization**

*   Right-click chart → **Clone**.
*   Change X-axis to **pickup\_datetime**.
*   Change color for better distinction.

***

#### **Step 6: Create Scatterplot**

*   Add new visualization → Select **Scatter**.
*   Configure:
    *   X-axis → trip\_distance
    *   Y-axis → fare\_amount
    *   Color → day\_of\_week

***

#### **Step 7: Add Filters**

*   Click **Add Filter**:
    *   **Date range filter** → tpep\_pickup\_datetime.
    *   **Dropdown filter** → dropoff\_zip.
*   Clone filter → Change to pickup\_zip.

***

#### **Step 8: Arrange Layout**

*   Drag and resize charts and filters on the canvas.

***

#### **Step 9: Publish & Share**

*   Click **Publish** → Choose **Share with data permissions**.
*   Add users/groups → Set permissions.
*   Copy link → Share with your team.

***

### ✅ **Quick Memory Tip**

Think of **Dashboard Creation** as:
**Create → Data → Visualize → Filter → Arrange → Publish & Share**

***

### ✅ **What is Databricks Assistant?**

*   It’s an **AI helper** inside Databricks that lets you **create charts using natural language**.
*   Instead of manually configuring everything, you just **type what you want**, and it builds the visualization.
*   Example:
        Bar chart of average fare amount over hourly dropoff time

***

### ✅ **How It Works**

1.  **Start with a Dashboard**
    *   Open your dashboard from the **Dashboards page**.
    *   Make sure you have at least **one dataset defined** in the **Data tab**.

2.  **Go to Canvas**
    *   Click **Canvas tab** → Add a **visualization widget**.

3.  **Type Your Prompt**
    *   In the widget, type what you want:
            Show average sales per week by region as a bar chart
    *   Press **Enter** → Assistant generates the chart.

4.  **Accept or Regenerate**
    *   If the chart looks good → Click **Accept**.
    *   If not → Click **Regenerate** or edit your prompt and retry.

***

### ✅ **Edit the Chart**

*   Two ways:
    *   **Use Assistant again** → Click the **Sparkle icon** and type a new prompt.
    *   **Use Configuration Panel** → Change X-axis, Y-axis, colors, etc.

***

### ✅ **Supported Chart Types**

*   Area, Bar, Line, Pie, Scatter, Table, Counter.

***

### ✅ **Tips for Better Results**

*   **Be specific**:
    *   Mention chart type: “Bar chart”
    *   Mention fields: “Average fare amount by day of week”
*   Assistant uses **metadata only** (table & column names), not actual row values.
*   Always **verify the chart** using the configuration panel.

***

### ✅ **Examples of Prompts**

*   “Bar chart of average fare amount over hourly dropoff time”
*   “Scatter plot of trip distance vs fare amount by day of week”
*   “Line chart of weekly sales trend”

***

### ✅ **Quick Memory Trick**

Think of it as:
**Prompt → Generate → Verify → Edit**

Here’s a **simple and clear explanation of query-based parameters** so you can understand and remember easily:

***

### ✅ **What Are Query-Based Parameters?**

*   A **parameter** is like a **placeholder** in your SQL query that you can change dynamically.
*   Instead of hardcoding a value (e.g., `WHERE c_mktsegment = 'BUILDING'`), you use:
        WHERE c_mktsegment = :segment
*   `:segment` is the parameter. You can change its value without editing the SQL query.

***

### ✅ **Why Use Parameters?**

*   Makes dashboards **interactive**.
*   Lets viewers **choose values** (like from a dropdown) instead of typing SQL.
*   Example:
    *   Parameter: `:segment`
    *   Viewer selects **AUTOMOBILE** → Dashboard updates automatically.

***

### ✅ **Steps to Use Query-Based Parameters**

#### **Step 1: Create Dataset with Parameter**

*   Write SQL:
    ```sql
    SELECT * FROM samples.tpch.customer
    WHERE c_mktsegment = :segment
    ```
*   Default value: Type **BUILDING** in the parameter box.
*   Run query → Dataset shows only rows for **BUILDING**.

***

#### **Step 2: Add Visualization**

*   Go to **Canvas tab**.
*   Add a chart:
    *   X-axis → `c_nationkey`
    *   Y-axis → `c_acctbal` (Transform: SUM)
*   Chart now shows data filtered by **BUILDING**.

***

#### **Step 3: Add Filter Widget**

*   Add **Filter** to dashboard.
*   Choose **Single value** filter type.
*   Link it to **parameter `segment`**.
*   Rename filter → **Segment**.
*   Now viewers can type a value (e.g., AUTOMOBILE) → Dashboard updates.

***

#### **Step 4: Make It Easy (Dropdown List)**

*   Create another dataset:
    ```sql
    SELECT DISTINCT c_mktsegment FROM samples.tpch.customer
    ```
*   Rename it **Segment choice**.
*   Update filter:
    *   Use **Segment choice** dataset.
    *   Field: `c_mktsegment`.
*   Now filter shows a **dropdown list** of all segments.

***

### ✅ **Quick Memory Tip**

Think of it as:
**Parameter = Dynamic Value → Filter = User Control → Dropdown = Easy Selection**

***

### ✅ **Real Example**

*   Parameter: `:segment`
*   Viewer picks **AUTOMOBILE** from dropdown → Dashboard updates charts instantly.
