## ✅ **Lakeflow Components**

1.  **Lakeflow Connect**
    *   **Purpose:** Ingest data from external sources (databases, SaaS apps, APIs) into the Lakehouse.
    *   **How:** Point-and-click connectors, high-throughput ingestion.
    *   **Use Case:** Load raw data quickly without writing custom ingestion code.

2.  **Lakeflow Declarative Pipelines (previously know as : DLT – Delta Live Tables)**
    *   **Purpose:** Transform and curate data using **declarative ETL**.
    *   **How:** Define tables and transformations in **SQL or Python**, pipeline config in **YAML**.
    *   **Features:** Automatic dependency resolution, incremental updates, data quality checks.
    *   **Use Case:** Production-grade ETL with minimal orchestration logic.

3.  **Lakeflow Jobs**
    *   **Purpose:** Orchestrate workflows (notebooks, pipelines, ML tasks).
    *   **How:** Imperative orchestration with triggers, branching, retries.
    *   **Use Case:** Combine multiple tasks into a scheduled workflow.

***

***

### ✅ **Lakeflow Jobs (Imperative Approach)**

*   You have **multiple notebooks** (think of them like big notebooks full of detailed instructions and code).
*   Each notebook contains **step-by-step code** for tasks like:
    *   Reading data
    *   Cleaning data
    *   Transforming data
    *   Loading data into Delta tables
*   To run these notebooks in the right order (or in parallel), you use **Lakeflow Jobs**.
*   This is called **imperative** because:
    *   You **tell the system exactly what to do and in what order**.
    *   Example: “Run Notebook 1 → then Notebook 2 → then Notebook 3.”
*   **Analogy:** It’s like cooking at home with a recipe:
    *   You write down every step and follow it yourself.
    *   If you have 5 dishes, you decide which one to cook first and which one next.

***

### ✅ **Lakeflow Declarative Pipeline**

*   Here, you **don’t write detailed code in notebooks**.
*   Instead, you **declare what you want to happen**, and the system figures out the steps.
*   Example:
    *   You say: “Ingest data from ADLS → Transform → Load into Delta table.”
    *   You don’t write the detailed Spark code for reading files, parsing formats, or handling schema.
*   It’s like **calling an API**:
    *   You say **what needs to happen**, not **how to make it happen**.
*   **Analogy:** Ordering food at a restaurant:
    *   You say “I want pizza,” and the chef handles all the steps behind the scenes.

***

### ✅ **Key Difference**

*   **Lakeflow Jobs (Imperative)**:
    *   You write **detailed code in notebooks**.
    *   You control the execution order.
    *   You manage orchestration manually.
*   **Lakeflow Declarative Pipeline**:
    *   You **don’t write detailed code**.
    *   You just declare the outcome.
    *   The system automatically handles orchestration and execution.

***

### **Visual Analogy**

*   **Imperative = You cook at home with a recipe (write all steps).**
*   **Declarative = You order food at a restaurant (just say what you want).**

## ✅ **Imperative vs Declarative**

*   **Imperative:**
    *   You tell the system **HOW** to do something step by step.
    *   Example: Python notebook with Spark code.
*   **Declarative:**
    *   You tell the system **WHAT** you want, and it figures out the execution.
    *   Example: DLT SQL:
        ```sql
        CREATE OR REFRESH STREAMING TABLE clean_sales AS
        SELECT * FROM raw_sales WHERE status = 'valid';
        ```

**Relation to OOP vs Procedural:**

*   Different concept. OOP vs Procedural is about **code structure**, while Imperative vs Declarative is about **execution style**.

*   **Complementary**:
    *   Use **DLT** for structured, automated ETL pipelines.
    *   Use **Notebooks** for ad-hoc analysis or complex custom logic.
*   **Lakeflow Jobs** can orchestrate **both notebooks and DLT pipelines**.

*   **Pipeline configuration:** YAML (clusters, libraries, expectations).

## ✅ **Why use DLT instead of just notebooks + Jobs?**

*   Jobs orchestrate tasks but **don’t manage ETL intelligence**.
*   DLT automates:
    *   Dependency graph
    *   Incremental updates
    *   Streaming vs batch handling
    *   Data quality enforcement
*   This reduces manual orchestration and errors.

## ✅ **What is a Connector in Databricks?**

A **connector** in Databricks is a prebuilt integration that allows you to **read, write, or stream data between Databricks and external systems** without writing low-level API or JDBC code.

*   It abstracts complexity like authentication, schema handling, and performance tuning.
*   Examples: **Databricks Lakeflow Connect**, JDBC connectors, Delta Sharing connectors.

***

## ✅ **When Was It Introduced?**

*   **Native connectors** (JDBC, Spark connectors) have existed for years.
*   **Lakeflow Connect** (managed connectors for ingestion) was introduced in **2024** as part of Databricks Lakeflow to simplify **high-throughput ingestion from SaaS and enterprise systems**.
*   Before this, users relied on:
    *   **Custom Spark jobs**
    *   **Third-party ETL tools** (Fivetran, Informatica, ADF)
    *   **Manual API integration**

***

## ✅ **Types of Connectors in Databricks**

1.  **Lakeflow Managed Connectors**
    *   Purpose: High-speed ingestion from external systems (Salesforce, SAP, ServiceNow).
    *   Features: Schema evolution, incremental sync, point-and-click setup.
    *   Best for: **Streaming or batch ingestion at scale** without coding.

2.  **Standard Spark/JDBC Connectors**
    *   Purpose: Connect to relational databases or file systems using Spark.
    *   Features: Full control via code, flexible but manual.
    *   Best for: **Custom ETL logic** or when managed connectors don’t exist.

3.  **Delta Sharing Connectors**
    *   Purpose: Share data securely across organizations.
    *   Best for: **Data collaboration** without copying data.

4.  **Partner Connect (Fivetran, ADF, Informatica)**
    *   Purpose: Use external ETL tools integrated with Databricks.
    *   Best for: Teams already invested in **Fivetran or ADF** for orchestration and governance.

***

## ✅ **When to Use Which Connector?**

*   **Lakeflow Managed Connector**:
    *   If you need **fast, automated ingestion** from SaaS or enterprise apps.
    *   Example: Pulling Salesforce data into Delta tables with schema evolution.

*   **Standard Spark/JDBC Connector**:
    *   If you need **custom transformations** or advanced Spark logic.
    *   Example: Complex joins, transformations during ingestion.

*   **Partner Connect (Fivetran, ADF)**:
    *   If your organization already uses these tools for **data movement and governance**.
    *   Example: ADF for hybrid pipelines, Fivetran for SaaS data sync.

*   **Delta Sharing**:
    *   If you need **secure data sharing** across business units or partners without duplication.

***

### ✅ **Why use Fivetran or ADF with Databricks?**

*   **Fivetran**:
    *   Great for **fully managed SaaS ingestion** (hundreds of connectors).
    *   Handles schema drift and incremental sync automatically.
*   **Azure Data Factory (ADF)**:
    *   Ideal for **hybrid orchestration** (on-prem + cloud).
    *   Strong in **data movement and scheduling** across Azure ecosystem.

***

### ✅ **Big Picture**

*   **Lakeflow Connect** → Simplifies ingestion inside Databricks.
*   **DLT (Declarative Pipelines)** → Automates ETL logic.
*   **Lakeflow Jobs** → Orchestrates workflows.
*   **Fivetran/ADF** → External orchestration or ingestion when enterprise standards require.

### ✅ **Imperative Approach (Lakeflow Jobs with Notebooks)**

You write **detailed code in multiple notebooks** and orchestrate them manually.

#### **Notebook 1: Ingest Data**

```python
# Read raw CSV files from ADLS
df = spark.read.format("csv") \
    .option("header", "true") \
    .load("abfss://mycontainer@myaccount.dfs.core.windows.net/raw/customers")

# Save as Delta
df.write.format("delta").mode("overwrite").save("/delta/raw/customers")
```

#### **Notebook 2: Transform Data**

```python
# Read Delta table
df = spark.read.format("delta").load("/delta/raw/customers")

# Apply transformation
df_clean = df.filter(df["status"] == "active")

# Save transformed data
df_clean.write.format("delta").mode("overwrite").save("/delta/clean/customers")
```

#### **Notebook 3: Create Final Table**

```python
spark.sql("""
CREATE TABLE customers_final
USING DELTA
LOCATION '/delta/clean/customers'
""")
```

**Lakeflow Job orchestration:**

*   Run Notebook 1 → Notebook 2 → Notebook 3 in sequence or parallel.

***

### ✅ **Declarative Approach (Lakeflow Pipeline)**

You **don’t write detailed Spark code**. You just declare what needs to happen.

#### **Pipeline YAML Example**

```yaml
pipeline:
  name: customer_pipeline
  steps:
    - ingest:
        source: adls://mycontainer/raw/customers
        format: csv
        options:
          header: true
    - transform:
        filter: status == 'active'
    - load:
        target: delta_table
        name: customers_final
```

**What happens:**

*   System automatically generates Spark code for reading, filtering, and writing.
*   Handles orchestration internally.

***

✅ **Key Difference in Syntax**

*   **Imperative:** You write full Spark code in notebooks and control execution.
*   **Declarative:** You write a simple pipeline definition; system does the heavy lifting.

### ✅ **What is a Connector in Databricks?**

A **connector** is a way to bring data from an external source (like Salesforce, SQL Server, or an API) into Databricks for processing and analytics.  
Think of it as a **bridge** between your data source and Databricks.

***

### ✅ **Managed Connector**

*   **What it is**: A ready-made, fully automated pipeline provided by Databricks.
*   **Key Features**:
    *   Runs on **serverless compute** (you don’t manage infrastructure).
    *   Handles **schema changes, retries, and incremental loads automatically**.
    *   Integrated with **Unity Catalog** for governance.
*   **Use Case**:
    *   You want to ingest data from **Salesforce** into Databricks without writing complex ETL code.
*   **Example**:
    *   You select **Salesforce Managed Connector** in Databricks UI.
    *   Configure credentials → choose tables → Databricks automatically creates streaming tables in Delta format.
    *   No need to manage Spark jobs or clusters.

***

### ✅ **Standard Connector**

*   **What it is**: A connector that gives you flexibility to build custom ingestion pipelines using **Spark Structured Streaming** or **Lakeflow Declarative Pipelines**.
*   **Key Features**:
    *   Runs on **your Databricks cluster** (not serverless).
    *   You write code or configure transformations.
    *   You manage schema evolution and error handling.
*   **Use Case**:
    *   You need to ingest data from a **custom API** or a source not supported by Managed connectors.
*   **Example**:
    *   You use **Spark Structured Streaming** to read JSON from an API endpoint.
    *   Apply custom transformations (e.g., flatten nested fields).
    *   Write to Delta tables manually.

***

### ✅ **Quick Decision Rule**

*   **Managed** → When source is common (Salesforce, ServiceNow, SQL Server) and you want **speed + simplicity**.
*   **Standard** → When source is **custom** or you need **complex transformations**.

***

### ✅ **Visual Analogy**

*   **Managed Connector** = “Uber” → You just book and ride; everything else is handled.
*   **Standard Connector** = “Own Car” → You drive, maintain, and customize as you like.

### ✅ **What are we doing?**

We want to **bring data into Databricks**. Think of Databricks as a big notebook where you can store and analyze data.  
To get data into it, we use **connectors**. A connector is like a **pipe** that moves water (data) from one tank (source) to another (Databricks).

***

### ✅ **Two Main Ways to Move Data**

1.  **Batch Ingestion**

    *   Imagine you take a photo of your entire WhatsApp chat every night and save it.
    *   That’s **batch**: you grab a bunch of data at once, usually on a schedule (like every hour or day).
    *   **Traditional Batch**: Every time, you copy **everything** again—even old messages.
    *   **Incremental Batch**: You only copy **new messages** since the last time.
        *   Faster and saves effort because you don’t re-copy old stuff.

    **Example in real life**:

    *   Traditional: Download your entire Instagram feed every day.
    *   Incremental: Only download new posts since yesterday.

***

2.  **Streaming Ingestion**
    *   Imagine watching a live cricket match instead of waiting for the highlights.
    *   Data flows **continuously** as it happens—great for real-time stuff like stock prices or sensor data.
    *   **Micro-batch**: Instead of every second, you collect data every few seconds or minutes and process it in small chunks.
        *   Like checking WhatsApp every 2 minutes instead of every second.

***

### ✅ **Why does this matter?**

*   **Batch** is good for things that don’t change super fast (like daily sales reports).
*   **Streaming** is good for things that change all the time (like live game scores or IoT sensors).

***

### ✅ **How does Databricks do it?**

*   For **batch**, Databricks has tools that say:
    *   “Create a table from this data” (SQL way).
    *   Or “Load this file” (Python way).
*   For **incremental batch**, it uses smart tools that only pick new stuff.
*   For **streaming**, it uses something called **Auto Loader**—think of it as a robot that keeps checking for new data and loads it instantly.

***

### ✅ **Simple Analogy**

*   **Traditional Batch** = Copying your entire photo album every day.
*   **Incremental Batch** = Copying only new photos since yesterday.
*   **Streaming** = Watching a live video instead of downloading clips later.

## ✅ **1. SQL: `CREATE TABLE AS SELECT`**

*   **What it does**: Creates a new table in Databricks by selecting data from an existing source (like a file or another table).
*   **When to use**: For **traditional batch ingestion**—loading all data at once.
*   **Example**:

```sql
CREATE TABLE sales_data AS
SELECT * FROM parquet.`/mnt/data/sales/2025/`;
```

✔ This reads all rows from the Parquet file and creates a table called `sales_data`.

***

## ✅ **2. Python: `spark.read.load()`**

*   **What it does**: Reads data from a file or directory into a Spark DataFrame.
*   **When to use**: Also for **traditional batch ingestion**.
*   **Example**:

```python
df = spark.read.format("parquet").load("/mnt/data/sales/2025/")
df.show()
```

✔ This loads all data from the path into a DataFrame.

***

## ✅ **3. SQL: `COPY INTO`**

*   **What it does**: Loads data **incrementally** into a Delta table. It only picks **new files** since the last load.
*   **When to use**: For **incremental batch ingestion**—efficient for large datasets.
*   **Example**:

```sql
COPY INTO sales_delta
FROM '/mnt/data/sales/'
FILEFORMAT = PARQUET;
```

✔ This copies only new files into `sales_delta`.

***

## ✅ **4. Python: `spark.readStream` (Auto Loader with timed trigger)**

*   **What it does**: Reads data **incrementally** using Auto Loader, but in **micro-batches** (every few minutes).
*   **When to use**: For **incremental ingestion with near real-time updates**.
*   **Example**:

```python
df = (spark.readStream
      .format("cloudFiles")
      .option("cloudFiles.format", "parquet")
      .load("/mnt/data/sales/"))

df.writeStream.format("delta").option("checkpointLocation", "/mnt/checkpoints/").start("/mnt/delta/sales/")
```

✔ Auto Loader detects new files and loads them automatically.

***

## ✅ **5. Declarative Pipelines: `CREATE OR REFRESH STREAMING TABLE`**

*   **What it does**: Creates a **streaming table** that automatically refreshes as new data arrives.
*   **When to use**: For **incremental or streaming ingestion** without writing complex code.
*   **Example**:

```sql
CREATE OR REFRESH STREAMING TABLE sales_stream
AS SELECT * FROM parquet.`/mnt/data/sales/`;
```

✔ This keeps `sales_stream` updated with new files.

***

## ✅ **6. Python: `spark.readStream` (Auto Loader with continuous trigger)**

*   **What it does**: Reads data **continuously** (true streaming, not micro-batch).
*   **When to use**: For **real-time ingestion** (e.g., IoT sensors, Kafka).
*   **Example**:

```python
df = (spark.readStream
      .format("cloudFiles")
      .option("cloudFiles.format", "json")
      .load("/mnt/data/live/"))

df.writeStream.trigger(continuous="true").format("delta").start("/mnt/delta/live/")
```

✔ Continuous trigger means data flows in **as soon as it arrives**.

***

## ✅ **7. Declarative Pipelines (trigger mode continuous)**

*   **What it does**: Same as above but using SQL declarative syntax.
*   **Example**:

```sql
CREATE OR REFRESH STREAMING TABLE live_stream
AS SELECT * FROM json.`/mnt/data/live/`
OPTIONS (triggerMode = 'continuous');
```

### **What is Delta Lake?**

Think of Delta Lake as a **smart upgrade** for your data lake. Normally, a data lake is just a big folder with files (like a messy hard drive). Delta Lake makes it organized and powerful.

***

### **How does it work?**

*   Inside Delta Lake, you have **Delta Tables**.
*   These tables store data as **Parquet files** (just a fancy way to store data efficiently).
*   Delta adds **log files** (in JSON format) next to those data files.  
    These logs are like a **history book** that records every change you make.

***

### **Why are logs important?**

*   Every time you **insert, update, or delete** something, Delta writes it in the log.
*   This means:
    *   Your table always stays consistent.
    *   You can **time travel** (go back to an older version of your data if needed).

***

### **Why is this better than old data lakes?**

Before Delta:

*   If you wanted to change data, you had to **recreate the whole file manually**. Painful!
    With Delta:
*   You just run a command like `UPDATE` or `DELETE`. Easy and fast.

***

### **Cool Features of Delta Lake**

1.  **ACID transactions**  
    → Multiple people can read/write at the same time without messing things up.
2.  **DML support**  
    → You can use SQL commands like `INSERT`, `UPDATE`, `DELETE`, `MERGE`.
3.  **Time Travel**  
    → Go back to old versions for audits or recovery.
4.  **Schema enforcement & evolution**  
    → Keeps data clean but lets you change structure when needed.
5.  Works for **batch and streaming** data.
6.  **Open source** and scalable.

***

✅ In short: **Delta Lake = Organized, reliable, and easy-to-manage data lake with superpowers like time travel and ACID transactions.**

### **What is CTAS (Create Table As Select)?**

Imagine you have a bunch of files (CSV, JSON, Parquet) sitting in **cloud storage** (like Azure ADLS, AWS S3).  
You want to turn them into a **Delta table** so you can use cool features like **time travel, ACID transactions, schema enforcement**.

**CTAS** is like saying:  
“Take these files → Read them → Make a new Delta table.”

***

### **How does it work?**

*   You use **`CREATE TABLE new_table AS SELECT ... FROM read_files(...)`**
*   `read_files()` is the function that **reads your files** and converts them into a table format.

***

### **Key Points**

✔ Supports many file types: **CSV, JSON, XML, Parquet, Avro, ORC**  
✔ Automatically detects file format and schema  
✔ You can **specify options** (like delimiter for CSV)  
✔ Works for **batch ingestion** and **streaming ingestion** (with Auto Loader)

***

### **Simple Coding Examples**

#### ✅ **Example 1: Read CSV and Create Delta Table**

```sql
CREATE TABLE sales_data AS
SELECT *
FROM read_files(
  'abfss://mycontainer@myaccount.dfs.core.windows.net/raw/sales/',
  format => 'csv',
  header => 'true',
  delimiter => ','
);
```

**What this does:**

*   Reads all CSV files from the given path
*   Creates a Delta table called `sales_data`

***

#### ✅ **Example 2: Read JSON Files**

```sql
CREATE TABLE customer_info AS
SELECT *
FROM read_files(
  'abfss://mycontainer@myaccount.dfs.core.windows.net/raw/customers/',
  format => 'json'
);
```

**What this does:**

*   Reads JSON files
*   Creates a Delta table called `customer_info`

***

#### ✅ **Example 3: Auto Detect Format**

```sql
CREATE TABLE auto_detect_table AS
SELECT *
FROM read_files(
  'abfss://mycontainer@myaccount.dfs.core.windows.net/raw/data/'
);
```

**What this does:**

*   Automatically detects file format
*   Infers schema
*   Creates Delta table

***

#### ✅ **Example 4: Streaming with Auto Loader**

```sql
CREATE STREAMING TABLE streaming_sales AS
SELECT *
FROM read_files(
  'abfss://mycontainer@myaccount.dfs.core.windows.net/raw/streaming_sales/',
  format => 'csv'
);
```

**What this does:**

*   Creates a **streaming Delta table**
*   Keeps ingesting new files as they arrive

### **What is COPY INTO?**

Imagine you have a **folder in the cloud** (like a big box) where new files keep coming every day.  
You want to **move those files into a Delta table** so you can use them easily.

**COPY INTO** is like saying:  
“Take all the new files from that box and put them into my table. If I already added some files before, skip them.”

***

### **Why is it cool?**

*   It **only adds new files** (doesn’t reload old ones).
*   Works with many file types: **CSV, JSON, Parquet, XML**.
*   You can control:
    *   **FORMAT\_OPTIONS** → How to read files (e.g., header, delimiter).
    *   **COPY\_OPTIONS** → Behavior (e.g., allow schema changes, force reload).

***

### ✅ **Step 1: Create an Empty Table**

```sql
CREATE TABLE new_table;
```

*(This makes an empty Delta table. You can also define columns if you want.)*

***

### ✅ **Step 2: Copy Files from Cloud Storage**

#### Example for CSV files:

```sql
COPY INTO new_table
FROM 'abfss://mycontainer@myaccount.dfs.core.windows.net/raw/sales/'
FILEFORMAT = CSV
FORMAT_OPTIONS('header' = 'true', 'delimiter' = ',')
COPY_OPTIONS('mergeSchema' = 'true');
```

**What this does:**

*   Reads all CSV files from the folder.
*   Adds them to `new_table`.
*   If new files come tomorrow, run this again → It will only add the new ones.

***

#### Example for JSON files:

```sql
COPY INTO new_table
FROM 'abfss://mycontainer@myaccount.dfs.core.windows.net/raw/customers/'
FILEFORMAT = JSON
COPY_OPTIONS('force' = 'false'); -- skip already loaded files
```

***

#### Example for Parquet files:

```sql
COPY INTO new_table
FROM 'abfss://mycontainer@myaccount.dfs.core.windows.net/raw/parquet_data/'
FILEFORMAT = PARQUET;
```

***

### ✅ **Think of it like this:**

*   **CTAS** = “Make a new shelf and put all groceries on it.”
*   **COPY INTO** = “Keep adding new groceries to the shelf without touching old ones.”

***

✅ How COPY INTO Works

FROM → Path to cloud storage folder.
FILEFORMAT → Type of file (CSV, JSON, PARQUET, XML, AVRO, ORC).
FORMAT_OPTIONS() → How to read files (header, delimiter, etc.).
COPY_OPTIONS() → Behavior:

mergeSchema = true → Allow schema evolution.
force = true → Reload files even if already loaded.

## ✅ **What is Auto Loader?**

*   Auto Loader **automatically detects and ingests new files** from cloud storage into Delta tables.
*   Works in **incremental batch** or **streaming mode**.
*   **No extra setup**—just point to your folder and Auto Loader handles new files.
*   Built on **Spark Structured Streaming**, so it’s fast and scalable (billions of files).
*   Supports **Python and SQL**, and integrates with **Lakeflow Declarative Pipelines**.

***

## ✅ **Why Use Auto Loader?**

*   Ideal for **continuously growing datasets** (new files keep arriving).
*   Handles **schema inference and evolution** automatically.
*   **Idempotent** → Won’t reload files already processed.
*   Perfect for **real-time streaming** or **incremental batch ingestion**.

***

## ✅ **Python Example**

```python
# Read streaming data from cloud storage using Auto Loader
df = (spark.readStream
      .format("cloudFiles")                # Enable Auto Loader
      .option("cloudFiles.format", "json") # File type (CSV, JSON, Parquet, etc.)
      .option("cloudFiles.schemaLocation", "/schema/checkpoint") # Track schema
      .load("/Volumes/my_catalog/my_volume/raw/customers"))       # Path to files

# Write to Delta table with checkpointing
(df.writeStream
   .option("checkpointLocation", "/checkpoints/customers") # Track progress
   .trigger(processingTime="5 seconds")                    # Trigger interval
   .toTable("my_catalog.my_db.customers_final"))           # Delta table name
```

✅ **Key points to remember:**

*   `format("cloudFiles")` → Turns on Auto Loader.
*   `cloudFiles.format` → File type.
*   `cloudFiles.schemaLocation` → Where schema evolution info is stored.
*   `checkpointLocation` → Ensures fault tolerance and incremental processing.
*   `.trigger(processingTime="X")` → Controls how often to process new files.
*   `.toTable()` → Writes to Delta table.

***

## ✅ **SQL Example (Declarative Pipeline)**

```sql
CREATE OR REFRESH STREAMING TABLE customers_final
AS SELECT *
FROM STREAM read_files(
  '/Volumes/my_catalog/my_volume/raw/customers',
  format => 'json'
)
```

✅ **What happens:**

*   `STREAM` keyword → Enables streaming ingestion.
*   `read_files()` → Reads files from cloud storage.
*   Auto Loader runs behind the scenes to detect new files and load them incrementally.

***

## ✅ **Quick Memory Tip**

*   **CTAS** → One-time batch load (all files).
*   **COPY INTO** → Incremental batch (skip old files).
*   **Auto Loader** → Continuous ingestion (batch or streaming) with minimal setup.
