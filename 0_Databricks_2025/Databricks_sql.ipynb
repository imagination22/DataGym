### ✅ What is **Information Schema**?

Think of **Information Schema** like a **big library catalog** that tells you **what books (tables), chapters (columns), and rules (constraints)** exist in your data system.

*   It **doesn’t store your actual data**.
*   It **stores information about your data** (metadata).
*   It works with **Unity Catalog**, which is like the main librarian that organizes everything.

***

### ✅ Why do we need it?

If you want to know:

*   **What tables exist?**
*   **Who owns them?**
*   **What columns are inside?**
*   **What permissions people have?**

You ask **Information Schema** instead of checking every table manually.

***

### ✅ How does it work?

*   It gives you **views** (special tables) that describe things like:
    *   **TABLES** → All tables and views.
    *   **COLUMNS** → All columns in those tables.
    *   **SCHEMATA** → All schemas (folders for tables).
    *   **PRIVILEGES** → Who can access what.
    *   **TAGS** → Labels applied to tables or columns.

***

### ✅ Example Questions You Can Ask

*   **Show me all tables created today:**
    ```sql
    SELECT table_name, table_owner
    FROM system.information_schema.tables
    WHERE datediff(now(), last_altered) < 1;
    ```
*   **Show me all columns that are DOUBLE type:**
    ```sql
    SELECT table_name, column_name
    FROM information_schema.columns
    WHERE data_type = 'DOUBLE';
    ```

***

### ✅ Important Notes

*   It only works with **Unity Catalog** (not old Hive metastore).
*   Use **filters** like `WHERE table_schema = 'default'` to avoid slow queries.
*   LIMIT doesn’t make queries faster (it just cuts results).

***

### ✅ Memory Trick

Think of **Information Schema** as:

*   **A map of your data world** → It tells you what exists and where.
*   **Not the actual treasure** → It doesn’t hold data, just directions.

## ✅ **What is Information Schema?**

It’s like a **directory of metadata** for your Unity Catalog environment.  
Instead of guessing what tables, columns, or permissions exist, you query these views to **discover structure, security, and relationships**.

***

## ✅ **Grouped Explanation of Views**

### **1. Catalog-Level Views**

*   **CATALOGS** → Lists all catalogs (top-level containers for schemas and tables).  
    **Use when:** You want to see all catalogs in your metastore.
*   **CATALOG\_PRIVILEGES** → Shows who has access to catalogs.  
    **Use when:** Auditing permissions at the catalog level.
*   **CATALOG\_TAGS** → Tags applied to catalogs.  
    **Use when:** Searching or filtering catalogs by tags.
*   **CATALOG\_PROVIDER\_SHARE\_USAGE** → Shows external shares mounted on catalogs.  
    **Use when:** Tracking data sharing across catalogs.

***

### **2. Schema-Level Views**

*   **SCHEMATA** → Lists schemas inside catalogs.  
    **Use when:** You need to see all schemas in a catalog.
*   **SCHEMA\_PRIVILEGES** → Who can access schemas.  
    **Use when:** Permission audits.
*   **SCHEMA\_TAGS** → Tags applied to schemas.
*   **SCHEMA\_SHARE\_USAGE** → Which schemas are shared externally.

***

### **3. Table & Column-Level Views**

*   **TABLES** → All tables and views in a catalog.  
    **Use when:** Inventory of tables.
*   **VIEWS** → Details about views only.
*   **COLUMNS** → All columns in tables/views.  
    **Use when:** Find column names, types, or search for sensitive data.
*   **TABLE\_TAGS / COLUMN\_TAGS** → Tags applied to tables or columns.  
    **Use when:** Data classification or governance.
*   **TABLE\_PRIVILEGES** → Who can access tables/views.
*   **TABLE\_CONSTRAINTS / KEY\_COLUMN\_USAGE / REFERENTIAL\_CONSTRAINTS** → Info about primary keys, foreign keys.  
    **Use when:** Understanding relationships for data modeling.
*   **COLUMN\_MASKS / ROW\_FILTERS** → Security features applied to columns or rows.

***

### **4. Security & Access Views**

*   **METASTORE\_PRIVILEGES / METASTORES** → Who can access the metastore.
*   **CREDENTIALS / CREDENTIAL\_PRIVILEGES** → Info about credentials for external data.
*   **EXTERNAL\_LOCATIONS / EXTERNAL\_LOCATION\_PRIVILEGES** → External storage locations and permissions.
*   **RECIPIENTS / RECIPIENT\_TOKENS / RECIPIENT\_ALLOWED\_IP\_RANGES** → For Delta Sharing recipients and security.

***

### **5. Advanced / Special Views**

*   **ROUTINES / ROUTINE\_COLUMNS / ROUTINE\_PRIVILEGES** → Functions and their permissions.
*   **PARAMETERS** → Parameters for routines.
*   **PROVIDERS / SHARES / SHARE\_RECIPIENT\_PRIVILEGES** → Delta Sharing details.
*   **VOLUMES / VOLUME\_TAGS / VOLUME\_PRIVILEGES** → Info about volumes (Unity Catalog feature for files).

***

## ✅ **When & Why to Use**

*   **Data Discovery:** Find all tables, columns, and schemas without manual browsing.
*   **Governance & Compliance:** Check who has access to what (privileges).
*   **Security Audits:** Validate permissions on catalogs, schemas, tables.
*   **Data Modeling:** Understand constraints, keys, and relationships.
*   **Tagging & Classification:** Query tags for sensitive data tracking.
*   **External Sharing:** Monitor shares and recipients for Delta Sharing.

***

### ✅ Example Use Cases

*   **Find all tables created today:**
    ```sql
    SELECT table_name, table_owner
    FROM system.information_schema.tables
    WHERE datediff(now(), last_altered) < 1;
    ```
*   **List all columns with data type DOUBLE:**
    ```sql
    SELECT table_name, column_name
    FROM information_schema.columns
    WHERE data_type = 'DOUBLE';
    ```
*   **Audit who has access to a schema:**
    ```sql
    SELECT grantee, privilege_type
    FROM information_schema.schema_privileges
    WHERE schema_name = 'finance';
    ```

## ✅ **1. Why do we see `system.information_schema`?**

*   Normally, **Information Schema** exists inside a **catalog** (like `main.information_schema`).
*   **`system.information_schema`** is special:
    *   It gives you **metadata across all catalogs** in your metastore.
    *   Use it when you want a **global view** (e.g., all tables in all catalogs).
*   If you only care about one catalog (like `main`), use:
    ```sql
    SELECT * FROM main.information_schema.tables;
    ```
*   If you want everything across catalogs:
    ```sql
    SELECT * FROM system.information_schema.tables;
    ```

***

## ✅ **2. Categorizing Views by Importance**

### **Most Commonly Used (You’ll use these a lot)**

*   **TABLES** → Lists all tables and views.
*   **COLUMNS** → Lists all columns in tables/views.
*   **SCHEMATA** → Lists schemas in a catalog.
*   **TABLE\_PRIVILEGES** → Who can access tables.
*   **SCHEMA\_PRIVILEGES** → Who can access schemas.
*   **CATALOGS** → Lists catalogs.
*   **CATALOG\_PRIVILEGES** → Who can access catalogs.

***

### **Should Know (For governance & security audits)**

*   **TABLE\_TAGS / COLUMN\_TAGS / SCHEMA\_TAGS / CATALOG\_TAGS** → Data classification.
*   **ROW\_FILTERS / COLUMN\_MASKS** → Row-level and column-level security.
*   **CONSTRAINT\_TABLE\_USAGE / CONSTRAINT\_COLUMN\_USAGE / TABLE\_CONSTRAINTS** → Keys and constraints.
*   **EXTERNAL\_LOCATIONS / EXTERNAL\_LOCATION\_PRIVILEGES** → External storage permissions.
*   **METASTORE\_PRIVILEGES / METASTORES** → Metastore-level access.

***

### **Good to Know (Advanced or specialized use cases)**

*   **ROUTINES / ROUTINE\_COLUMNS / ROUTINE\_PRIVILEGES / PARAMETERS** → Functions and stored procedures.
*   **PROVIDERS / SHARES / SHARE\_RECIPIENT\_PRIVILEGES** → Delta Sharing details.
*   **RECIPIENTS / RECIPIENT\_TOKENS / RECIPIENT\_ALLOWED\_IP\_RANGES** → Sharing security.
*   **VOLUMES / VOLUME\_TAGS / VOLUME\_PRIVILEGES** → Unity Catalog Volumes.
*   **CREDENTIALS / CREDENTIAL\_PRIVILEGES** → External credentials.
*   **INFORMATION\_SCHEMA\_CATALOG\_NAME** → Returns catalog name (rarely needed).

***

### ✅ **Quick Rule**

*   **Most Common** → For data discovery and permissions.
*   **Should Know** → For compliance and governance.
*   **Good to Know** → For advanced features like sharing, routines, volumes.
