## ✅ **What is Databricks One?**

Databricks One is a **new, simplified user interface** designed for **business users**.  
It provides **a single entry point to interact with data and AI**—without dealing with technical concepts like clusters, notebooks, or queries.

Think of it as **a business-friendly dashboard for Databricks**.

***

## 💡 **Why Should I Care?**

*   **Easy access for non-technical users** → No coding required.
*   **Unified experience** → Dashboards, AI tools, and apps in one place.
*   **Natural language interaction** → Ask questions without writing SQL.
    *(Curiosity: Why not just use Power BI or Tableau?)*  
    **Answer:** Because Databricks One connects **directly to your lakehouse and AI models**, giving real-time insights and AI-powered answers.

***

## ⚙️ **How It Works (Key Features & Flow)**

### **1. Landing Page**

*   Shows **suggested assets**:
    *   Dashboards
    *   Genie spaces (AI-powered Q\&A)
    *   Databricks Apps
*   **Search bar** → Find assets or ask questions in natural language.

***

### **2. AI/BI Dashboards**

*   Visualize KPIs and metrics.
*   Create and share dashboards easily.
    *(Example: Sales performance dashboard for executives.)*

***

### **3. Ask Genie**

*   AI-powered Q\&A tool.

*   Genie interprets and returns results instantly.

***

### **4. Databricks Apps**

*   Custom apps combining **analytics, AI, and workflows**.
*   Example: A **forecasting app** for demand planning.

***

### **5. Access Control**

*   **Consumer Access** → View-only in Databricks One.
*   **Workspace or SQL Access** → Switch between Databricks One and Lakehouse UI.
    *(Curiosity: Why separate entitlements?)*  
    **Answer:** To keep business users focused on insights, not infrastructure.

## ✅ **What Does “Getting Identifiers” Mean?**

Every object in Databricks—**workspace, clusters, notebooks, jobs, dashboards, models, folders**—has a **unique ID and URL**.  
These identifiers are used for:

*   **Automation** (via REST API or CLI)
*   **Sharing links** with team members
*   **Troubleshooting jobs or clusters**

Think of IDs as **license plates for your Databricks assets**—unique and trackable.

***

## 💡 **Why Should I Care?**

*   **Automation** → APIs need IDs to manage objects programmatically.
*   **Collaboration** → Share stable URLs for notebooks, dashboards, or folders.
*   **Debugging** → Identify failed jobs or clusters quickly.
    *(Curiosity: Why not just use names?)*  
    **Answer:** Names can duplicate or change; IDs are **unique and permanent**.

***

## ⚙️ **How It Works (Key Identifiers & Where to Find Them)**

### **1. Workspace**

*   **Per-workspace URL format:**
        https://adb-<workspace-id>.<random>.azuredatabricks.net/
*   **Workspace ID:**  
    Appears after `adb-` in the URL.  
    *(Example: `adb-5555555555555555.19.azuredatabricks.net` → ID = `5555555555555555`)*

***

### **2. Cluster (Compute Resource)**

*   **Cluster ID:**  
    Found in the URL after `/clusters/`.  
    *(Example: `https://<instance>/compute/clusters/0130-201722-abcdefgh` → ID = `0130-201722-abcdefgh`)*

***

### **3. Notebook**

*   **Notebook URL format:**
        https://<instance>/?o=<workspace-ID>#notebook/<notebook-ID>
*   **Notebook ID:**  
    Appears after `#notebook/`.  
    *(Example: `1940481404050342`)*
*   **Cell URL:**  
    Adds `/command/<cell-ID>` for specific cells.

***

### **4. Folder**

*   **Directory ID:**  
    Retrieved via Workspace API:
    ```bash
    curl -X GET -d '{"path": "/Users/me@example.com/MyFolder"}' \
    https://<instance>/api/2.0/workspace/get-status
    ```
*   **Response:**
    ```json
    { "object_type": "DIRECTORY", "object_id": 123456789012345 }
    ```

***

### **5. Dashboard**

*   **Dashboard URL format:**
        https://<instance>/sql/dashboardsv3/<dashboard-ID>
*   **Dashboard ID:**  
    Example: `01ef9214fcc7112984a50575bf2b460f`

***

### **6. Job**

*   **Job URL format:**
        https://<instance>/jobs/<job-ID>?o=<workspace-ID>
*   **Job ID:**  
    Example: `5`

***

### **7. Model**

*   **Model ID:**  
    Retrieved via MLflow API:
    ```bash
    curl -X GET -d '{"name": "model_name"}' \
    https://<instance>/api/2.0/mlflow/databricks/registered-models/get
    ```
*   **Response:**
    ```json
    { "id": "ceb0477eba94418e973f170e626f4471" }
    ```

***

## 🔍 **Practical Use Cases**

*   **Automation:**  
    Use IDs in REST API calls to manage clusters, jobs, or models.
*   **Collaboration:**  
    Share stable URLs for notebooks or dashboards.
*   **Troubleshooting:**  
    Identify failed job runs using job ID.

***

## 🖥 **Examples**

### **Construct a Shareable Notebook URL**

Original:

    https://<instance>/?o=1111111111111111#notebook/2222222222222222/command/3333333333333333

Stable format:

    https://<instance>/?o=1111111111111111#workspace/Workspace/Users/user@example.com/team-git/notebooks/test-notebook

## ✅ **What is a Per-Workspace URL?**

A **Per-Workspace URL** is a **unique web address for each Databricks workspace**, introduced in April 2020.  
Format:

    adb-<workspace-id>.<random-number>.azuredatabricks.net

This replaces the **legacy regional URL** (like `eastus.azuredatabricks.net`) for better reliability and performance.

***

## 💡 **Why Should I Care?**

*   **More reliable** → Legacy URLs may break for new workspaces.
*   **Better performance** → Faster and more stable than regional URLs.
*   **Required for automation** → APIs and scripts should use per-workspace URLs.
    *(Curiosity: Why not keep regional URLs?)*  
    **Answer:** They are **deprecated**, less secure, and can cause compatibility issues.

***

## ⚙️ **How It Works**

*   Each workspace gets:
    *   **Unique URL** → `adb-<workspace-id>.<random>.azuredatabricks.net`
    *   **Workspace ID** → Appears after `adb-` in the URL.
*   Example:

<!---->

    https://adb-5555555555555555.19.azuredatabricks.net/
    Workspace ID = 5555555555555555

***

### **Launch Workspace**

*   From **Azure Portal**:
    *   Go to Databricks resource → Click **Launch Workspace**.
*   Or copy the **per-workspace URL** and paste in browser.

***

### **Get URL via Azure API**

*   Use **Workspaces - Get** endpoint:

```bash
GET https://management.azure.com/subscriptions/<sub-id>/resourceGroups/<rg>/providers/Microsoft.Databricks/workspaces/<workspace-name>?api-version=2023-02-01
```

*   Response:

```json
"properties": {
  "workspaceUrl": "adb-5555555555555555.19.azuredatabricks.net"
}
```

***

### **Update Scripts**

*   Replace hardcoded **regional URLs** with **per-workspace URLs**.
*   Store **\<per-workspace-url, API-token>** pairs for automation.

*(Curiosity: Will old scripts break?)*  
**Answer:** No, they still work for older workspaces—but updating is recommended.

***

### **Find Legacy Regional URL**

*   Run `nslookup` on per-workspace URL:

```bash
nslookup adb-<workspace-id>.<random>.azuredatabricks.net
```

*   Output shows canonical regional name (e.g., `eastus-c3.azuredatabricks.net`).

***

## 🔍 **Practical Use Cases**

*   **Automation** → REST API calls for clusters, jobs, models.
*   **CI/CD** → Use per-workspace URLs in deployment scripts.
*   **Troubleshooting** → Identify workspace by its unique ID.

***

### ✅ **Real-world Analogy**

Think of **per-workspace URLs like personalized home addresses**:

*   Old regional URLs = “City name only” (confusing for mail delivery).
*   Per-workspace URLs = “Exact street address” (precise and reliable).

***

✅ **Why This Matters for You**  
Switching to per-workspace URLs ensures **future-proof automation, better performance, and fewer errors**.

## **Databricks URL Cheat Sheet: Per-Workspace vs Legacy Regional**

### ✅ **What They Are**

*   **Legacy Regional URLs**
    *   Format: `https://<region>.azuredatabricks.net`
    *   Single URL per region for all workspaces.
    *   Example: `https://eastus.azuredatabricks.net`

*   **Per-Workspace URLs**
    *   Introduced April 2020 for Azure Databricks.
    *   Unique URL for each workspace.
    *   Format: `https://adb-<workspace-id>.<random>.azuredatabricks.net`
    *   Example: `https://adb-1234567890123456.7.azuredatabricks.net`    [\[learn.microsoft.com\]](https://learn.microsoft.com/en-us/azure/databricks/workspace/per-workspace-urls)

***

### ✅ **Why Per-Workspace URLs Are Better**

*   **Reliability & Performance**: Regional URLs are deprecated and less reliable; per-workspace URLs ensure better stability.
*   **Compatibility**: Required for new workspaces; legacy URLs may break features like workspace ID-based APIs.
*   **Security & Isolation**: Easier to manage access and automation per workspace.
*   **Future-Proof**: Aligns with Databricks Enterprise 2.0 architecture.    [\[learn.microsoft.com\]](https://learn.microsoft.com/en-us/azure/databricks/workspace/per-workspace-urls)

***

### ✅ **Examples**

*   **Legacy Regional**:
    *   `https://eastus.azuredatabricks.net`
    *   `https://us-east4.gcp.databricks.com`

*   **Per-Workspace**:
    *   `https://adb-1234567890123456.7.azuredatabricks.net`
    *   `https://8757561887652360.0.gcp.databricks.com`    [\[docs.databricks.com\]](https://docs.databricks.com/aws/en/workspace/workspace-details)

***

### ✅ **Migration Steps for Scripts & Automation**

1.  **Identify Hardcoded URLs**
    *   Search scripts for `<region>.azuredatabricks.net`.
2.  **Replace with Per-Workspace URLs**
    *   Use Azure API (`Workspaces - Get`) to fetch `workspaceUrl`.
3.  **Update Token Storage**
    *   Store `<per-workspace-url, api-token>` pairs instead of regional URLs.
4.  **Validate Automation**
    *   Test jobs, CI/CD pipelines, and integrations after update.
5.  **Optional**: Use `nslookup` on per-workspace URL to find legacy regional URL if needed.
    ```bash
    nslookup adb-<workspace-id>.<random>.azuredatabricks.net
    ```

 [\[learn.microsoft.com\]](https://learn.microsoft.com/en-us/azure/databricks/workspace/per-workspace-urls)

***

### ✅ **Quick Tips for Finding Workspace IDs & URLs**

*   **From Browser**:
    *   After login, check URL: `...?o=<workspace-id>`
    *   Example: `https://adb-1234567890123456.azuredatabricks.net/?o=6280049833385130`
*   **From Notebook**:
    ```python
    spark.conf.get("spark.databricks.workspaceUrl")
    spark.conf.get("spark.databricks.clusterUsageTags.orgId")  # Workspace ID
    ```
*   **Azure Portal**:
    *   Go to Databricks resource → Copy URL from “Launch Workspace”.
*   **API**:
    *   Use `GET /workspaces` endpoint; check `properties.workspaceUrl`.    [\[docs.databricks.com\]](https://docs.databricks.com/gcp/en/workspace/workspace-details)

***

#### **Pro Tip**: Avoid legacy URLs—they can break new features and slow down automation. Always migrate to per-workspace URLs for better reliability and compliance.
