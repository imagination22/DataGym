*   How do we keep it secure?

It works **across all workspaces** so everyone follows the same rules.

***

## ✅ **5: The Hierarchy – Organizing Data Like a Pro**

Unity Catalog uses a **three-level namespace**:

    catalog.schema.table

Here’s the breakdown:

*   **Metastore**: The **master index** of everything.
*   **Catalog**: Big container (like a warehouse).
*   **Schema**: Smaller section inside the warehouse.
*   **Data Objects**:
    *   **Tables** (structured data)
    *   **Views** (saved queries)
    *   **Volumes** (raw files like CSV, JSON)
    *   **Functions** (custom logic)

Example:

    finance.default.sales_data

***

## ✅ **6: Catalog Explorer – Your Map App**

Catalog Explorer is the **UI tool** that shows you:

*   All catalogs, schemas, tables, views, volumes.

*   Lets you manage permissions visually.

***

## ✅ **7: Types of Tables**

*   **Managed Tables**: Databricks owns the data. Drop the table = data gone.
*   **External Tables**: Point to cloud storage. Drop the table = data stays.
*   Others: Streaming tables, foreign tables, feature tables (ML).

***

## ✅ **8: Volumes & Views**

*   **Volumes**: For raw files (CSV, images). Governed by Unity Catalog.
*   **Views**: Read-only queries.
    *   **Dynamic views** (always fresh).
    *   **Materialized views** (cached for speed).

***

## ✅ **9: Functions**

Reusable logic stored in **Catalog + Schema**.  
Shared across your organization.

            └── Schema

### ✅ **1. What is Metastore**

*   
    A **metastore** is like a big index or directory that stores **metadata** (information about your data objects) and **access control rules**.
*   **Unity Catalog Metastore vs Hive Metastore**
    *   **Hive Metastore**: The old default for each Databricks workspace.
    *   **Unity Catalog Metastore**: The new standard with:
        *   Better **security**
        *   Centralized **governance**
        *   Auditing capabilities
        *   Advanced features for managing data and AI assets.
*   **Key Point**: Unity Catalog Metastore can be shared across multiple workspaces, unlike Hive which was tied to one workspace.

***

### ✅ **2. what is Catalog**

*  
    A **catalog** is the first level in the **three-level namespace**:
        catalog.schema.table
*   **Purpose**:
    *   Acts as a **container for schemas**.
    *   Helps **organize and isolate data** for different teams or projects.
*   **Example**:  
    `workspace` could be a catalog name.

***

### ✅ **3. What is Schema**

*   
    The second level in the namespace:
        catalog.schema.table
*   **Purpose**:
    *   Organizes data objects at a more granular level.
    *   Holds **tables**, **views**, and other data-bearing assets.
*   **Alias**:  
    In Databricks, **schema = database**.  
    Commands like `CREATE DATABASE` and `CREATE SCHEMA` mean the same thing.
*   **Example**:  
    `default` could be a schema name.

***

### ✅ **4. what are  Data Objects**

*   
    The third level in the namespace:
        catalog.schema.table
*   **Types**:
    *   **Tables**: Structured data (rows and columns).
    *   **Views**: Saved queries.
    *   **Volumes**: Raw files (CSV, JSON, images).
    *   **Functions**: Custom logic.
    *   **Models**: AI/ML models.
*   **Example**:  
    `sales_data` could be a table name.

So a fully qualified name looks like:

    workspace.default.sales_data

### ✅ **Unity Catalog – What It Does**

*   Unity Catalog is the **data governance system** in Databricks.
*   It controls **who can access what data and objects** (tables, views, volumes, notebooks, dashboards).
*   It ensures **security and compliance** by applying strict access rules.

***

### ✅ **Principle of Least Privilege (PoLP)**

*   This is a **security best practice**.
*   It means:
    *   Every user gets **only the permissions they need** to do their job.
    *   No extra access.
*   Why?
    *   Reduces risk of **data leaks** or **misuse**.
    *   Keeps your organization compliant and secure.

**Example:**  
If you’re a Data Analyst, you can read tables you need, but you won’t automatically get access to all tables in the company.

***

### ✅ **What It Means for You**

*   When you create a **table, view, or notebook**, you become its **owner**.
*   Owners can:
    *   Read and modify the object.
    *   Grant permissions to others.
*   Your teammates **do NOT automatically get access** unless:
    *   Permissions are set at the **group level** by admins.
    *   Or you **manually grant access**.

***

### ✅ **Principals (Identities) in Databricks**

*   **Users**: Real people (identified by email).
*   **Service Principals**: Non-human identities (e.g., API connections, automation jobs).
*   **Account Groups**: Groups of users/service principals for easier permission management.

***

### ✅ **Ownership and Permissions**

*   Every object (table, schema, catalog, notebook) has an **owner**.
*   Owners can grant permissions to:
    *   Other users
    *   Service principals
    *   Groups
*   Permissions **inherit down the hierarchy**:
    *   Catalog owner → can manage schemas and tables inside.
    *   Schema owner → can manage tables inside.

***

### ✅ **Permissions for Workspace Objects (Notebooks, Dashboards)**

*   **CAN VIEW**: See the object and its results.
*   **CAN RUN**: Execute the object (e.g., run a notebook).
*   **CAN EDIT**: Change the content (code, comments).
*   **CAN MANAGE**: Change permissions and delete the object.

***

### ✅ **Key Takeaways**

*   **Creating a table does NOT automatically share it with your team.**
*   You must **grant permissions** explicitly or rely on group-level policies.
*   Sharing a **notebook** is separate from sharing the **data** it uses.
*   Always follow **PoLP**: give only the access needed.

***



***

### ✅ **Permissions on Data Objects**

*   In **Unity Catalog**, permissions are **very specific** for different objects:
    *   **Catalogs** (top level)
    *   **Schemas** (inside catalogs)
    *   **Data Objects** (tables, views, volumes, models)
*   Why so detailed?  
    Because each type of object needs different levels of control. For example:
    *   A **table** might allow `SELECT` (read data) or `INSERT` (add data).
    *   A **volume** might allow `READ FILES` or `WRITE FILES`.
*   There are many permission types (too many to list here), but you can check the full list in <https://docs.databricks.com/aws/en/data-governance/unity-catalog/manage-privileges/privileges>.

***

### ✅ **Your Role in Security**

*   Databricks is built with **data security and governance in mind**.
*   As a **data analyst**, you:
    *   Need to know **what permissions you have**.
    *   Understand **how to share data safely**.
    *   Request access when you need it.

***

### ✅ **Sharing Objects**

*   **Notebooks and Queries**:
    *   You can share them with permissions like:
        *   **CAN VIEW** → See the notebook.
        *   **CAN RUN** → Execute it.
        *   **CAN EDIT** → Change code.
        *   **CAN MANAGE** → Change permissions or delete.
*   **Data Objects (Tables, Views, Volumes)**:
    *   If you are the **owner**, you can **GRANT privileges** to others.
    *   Example:
        ```sql
        GRANT SELECT ON TABLE my_table TO `data_analyst_group`;
        ```
    *   If you **don’t own** the object, you **won’t see it** in Catalog Explorer.

***

### ✅ **Important Points**

*   If your **data engineering team creates tables for you**, they must **share them with you**.
*   If you want to **collaborate on notebooks**, the owner must give you the right permissions.
*   Permissions are **not automatic**—they follow the **Principle of Least Privilege** (only what’s needed).

***

### ✅ **In Short**

*   **Unity Catalog = Fine-grained control** over who can do what.
*   **Owners** manage permissions for their objects.
*   **Groups** make sharing easier (instead of individual users).
*   Always **request access** if you can’t see what you need.

Databricks uses **Unity Catalog** to keep data organized and secure. Everything you work with—tables, views, files, models—are called **securable objects** because you can control who has access to them.

### ✅ **Hierarchy (Think of it like folders)**

1.  **Metastore**
    *   The top-level container (like the main library).
    *   Organizes everything using **three levels**:  
        `catalog.schema.table`

2.  **Catalog**
    *   First layer inside the metastore (like a big section in the library).
    *   Holds **schemas**.

3.  **Schema**
    *   Second layer (like a shelf in the section).
    *   Holds **tables and views**.

4.  **Data Objects**
    *   Third layer (like books on the shelf):
        *   **Table**: Actual data (rows and columns).
        *   **View**: A saved query (read-only).
        *   **Materialized View**: A view that stores results for faster access.
        *   **Metric View**: Special view for metrics.
        *   **Volume**: For raw files (.csv, .txt, .parquet).
        *   **Function**: Saved logic or ML model.
        *   **Model**: A machine learning model.
        *   **External Location**: Points to cloud storage.
        *   **Share/Recipient/Provider**: For Delta Sharing (sharing data with others).
        *   **Clean Room**: A secure space for collaboration without exposing raw data.

***

### ✅ **Permissions (Who can do what?)**

*   **Metastore**: Can create catalogs, shares, connections, etc.
*   **Catalog**: Can create schemas, tables, functions, volumes.
*   **Schema**: Can create tables, views, models.
*   **Table/View**: Can `SELECT` (read), `MODIFY` (change), `MANAGE` (control permissions).
*   **Volume**: Can read/write files.
*   **Functions/Models**: Can execute or manage.
*   **Shares**: Can grant access to recipients.

***

### ✅ **Easy Way to Remember**

*   **Metastore → Catalog → Schema → Table/View**  
    (Like Library → Section → Shelf → Book)
*   Permissions go from **broad to specific**:
    *   Metastore = Big powers (create catalogs).
    *   Catalog = Organize schemas.
    *   Schema = Organize tables/views.
    *   Table/View = Read or edit data.

***

### ✅ **Why So Many Permissions?**

*   To follow **Principle of Least Privilege**:
    *   Give people **only what they need**.
    *   Keeps data safe and prevents mistakes.

### 🎓 **Imagine This: A Big School Library**

*   **Unity Catalog** = The library security system.
*   **Privileges** = What you’re allowed to do in the library.
*   **Objects** = Sections, shelves, books, and lockers.

***

#### ✅ **Here’s the Fun Breakdown**

***

### **1. Big Boss Power**

*   **ALL PRIVILEGES** = You’re the **Principal**. You can do almost everything in the library (read, write, create shelves), but you **can’t change the security system** (MANAGE) or give outside people keys (EXTERNAL USE).

***

### **2. See & Access**

*   **BROWSE** = Peek at book titles without opening them.
*   **USE CATALOG** = Permission to enter the section (Science section).
*   **USE SCHEMA** = Permission to touch the shelf.
*   **ACCESS** = Special key to open a locked room (service credential).

***

### **3. Create Stuff**

*   **CREATE CATALOG** = Add a new section (Math section).
*   **CREATE SCHEMA** = Add a new shelf.
*   **CREATE TABLE / VIEW** = Add new books or summaries.
*   **CREATE FUNCTION / MODEL** = Add formulas or ML models (like cheat sheets).
*   **CREATE VOLUME** = Add a box for files.
*   **CREATE CONNECTION** = Connect to another school library.
*   **CREATE CLEAN ROOM** = Secret study room for group projects.

***

### **4. Control & Manage**

*   **MANAGE** = Act like the librarian: change permissions, rename, delete.
*   **MANAGE ALLOWLIST** = Decide which tools/scripts are allowed in the library.
*   **MODIFY** = Edit book content (update data).
*   **MODIFY CLEAN ROOM** = Rearrange the secret study room.
*   **APPLY TAG** = Put colorful stickers on books (tags).

***

### **5. Read & Write**

*   **SELECT** = Read the book.
*   **READ VOLUME** = Read files in a box.
*   **WRITE VOLUME** = Add or remove files in a box.
*   **READ FILES / WRITE FILES** = Directly read/write in external storage (like a locker outside school).

***

### **6. Execute & Refresh**

*   **EXECUTE** = Run a formula or ML model (like solving math problems).
*   **EXECUTE CLEAN ROOM TASK** = Run notebooks in the secret room.
*   **REFRESH** = Update a summary page (materialized view).

***

### **7. External Access**

*   **EXTERNAL USE LOCATION** = Temporary key to access lockers outside school.
*   **EXTERNAL USE SCHEMA** = Temporary key to access shelves from outside.

***

### ✅ **Memory Hack**

Think of **7 buckets**:

1.  **ALL** → Principal power
2.  **SEE** → Peek and enter
3.  **CREATE** → Build new stuff
4.  **CONTROL** → Act like librarian
5.  **READ/WRITE** → Books and boxes
6.  **EXECUTE** → Solve problems
7.  **EXTERNAL** → Outside access
