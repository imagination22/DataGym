Here’s a **detailed explanation of how to add and manage comments for data and AI assets in Unity Catalog, with examples**:

***

## ✅ **What Are Comments in Unity Catalog?**

*   Comments are **metadata annotations** for securable objects like:
    *   **Catalogs, Schemas, Tables, Volumes, AI Models**
    *   **Columns in tables**
*   Purpose:
    *   Improve **data discoverability**.
    *   Provide **context** for users browsing data assets.
*   Users with **BROWSE privilege** can view comments.

***

## ✅ **Permissions Required**

*   For **catalog, schema, volume, model, etc.**:
    *   Must be **owner** or have `MANAGE` privilege.
*   For **tables and columns**:
    *   Must be **owner** or have:
        *   `MODIFY` and `SELECT` on the table.
        *   `USE CATALOG` and `USE SCHEMA` on parent catalog/schema.
*   Compute must support **Unity Catalog** (SQL warehouse or UC-enabled cluster).

***

## ✅ **Ways to Add Comments**

You can add comments using:

1.  **Catalog Explorer (UI)**
    *   Navigate to **Data → Catalog**.
    *   Select object → Add/Edit comment in **Overview tab**.
    *   For columns: click **Add comment icon** next to column name.
    *   Supports **basic Markdown** (except images).
    *   AI-generated comments available for tables and columns.

2.  **SQL Commands**
    *   Use `COMMENT ON` or `ALTER` commands.
    *   Add comments during **CREATE** or **ALTER** operations.

***

## ✅ **Examples**

### **1. Add Comment When Creating a Table**

```sql
CREATE TABLE finance_catalog.sales_schema.orders (
    order_id STRING,
    customer_id STRING,
    amount DOUBLE COMMENT 'Transaction amount in USD',
    order_date DATE
)
COMMENT 'Orders table storing customer transactions';
```

***

### **2. Add Comment to an Existing Table**

```sql
COMMENT ON TABLE finance_catalog.sales_schema.orders IS 'Stores all customer orders with transaction details';
```

***

### **3. Add Comment to a Column**

```sql
ALTER TABLE finance_catalog.sales_schema.orders
ALTER COLUMN amount COMMENT 'Transaction amount in USD';
```

***

### **4. Add Comment to a Schema**

```sql
COMMENT ON SCHEMA finance_catalog.sales_schema IS 'Schema for all sales-related data assets';
```

***

### **5. Add Comment to a Volume**

```sql
COMMENT ON VOLUME finance_catalog.sales_schema.sales_volume IS 'Volume for storing raw sales files';
```

***

### **6. Edit Comment on a Table**

```sql
ALTER TABLE finance_catalog.sales_schema.orders
SET TBLPROPERTIES ('comment' = 'Updated description for orders table');
```

***

## ✅ **AI-Generated Comments**

*   Available for **tables and columns** in Catalog Explorer.
*   You can **accept**, **edit**, and **save** AI-suggested comments.

***

## ✅ **Best Practices**

*   Use **clear, descriptive comments** for better discoverability.
*   Include **business context** (e.g., KPIs, data source).
*   For sensitive data, note **masking or security rules** in comments.

Here’s a **complete explanation of applying tags to Unity Catalog securable objects with practical examples**:

***

## ✅ **What Are Tags?**

Tags are **key-value attributes** that help you:

*   Organize and classify data assets (e.g., sensitivity, department, project).
*   Improve **search and discovery** for tables and views.
*   Support **governance and compliance** by marking sensitive data.

> **Important:** Tags are stored as plain text and may replicate globally. **Never include sensitive information** in tag keys or values.

***

## ✅ **Supported Objects**

You can apply tags to:

*   **Catalogs**
*   **Schemas**
*   **Tables**
*   **Table columns**
*   **Volumes**
*   **Views**
*   **Models and model versions**

***

## ✅ **Tag Inheritance**

*   Tags applied to a **catalog** → inherited by all schemas and objects inside.
*   Tags applied to a **schema** → inherited by all objects inside.
*   **Column tags do not inherit** from table tags.

***

## ✅ **Governed Tags**

*   Enforce **standardized keys and values** across the organization.
*   Only users with **ASSIGN permission** can apply governed tags.
*   Example governed tag: `sensitivity = PII`.

***

## ✅ **Permissions Required**

To apply tags:

*   `APPLY TAG` on the object.
*   `USE SCHEMA` on parent schema.
*   `USE CATALOG` on parent catalog.
*   For governed tags: `ASSIGN` permission on the tag.

***

## ✅ **Constraints**

*   Max **50 tags per object**.
*   Max **1,000 column tags per table**.
*   Tag key length: **255 chars**, value length: **1,000 chars**.
*   Disallowed characters in keys: `. , - = / :`
*   No leading/trailing spaces.

***

## ✅ **Examples**

### **1. Add Tag to a Catalog**

```sql
SET TAG ON CATALOG finance_catalog `cost_center` = `hr`;
```

### **2. Remove Tag from a Catalog**

```sql
UNSET TAG ON CATALOG finance_catalog cost_center;
```

***

### **3. Add Tags to a Table**

```sql
ALTER TABLE finance_catalog.sales_schema.orders
SET TAGS ('sensitivity' = 'PII', 'department' = 'Finance');
```

### **4. Remove Tags from a Table**

```sql
ALTER TABLE finance_catalog.sales_schema.orders
UNSET TAGS ('sensitivity');
```

***

### **5. Add Tag to a Column**

```sql
ALTER TABLE finance_catalog.sales_schema.orders
ALTER COLUMN email
SET TAGS ('pii' = 'email');
```

### **6. Remove Tag from a Column**

```sql
UNSET TAG ON COLUMN finance_catalog.sales_schema.orders.email pii;
```

***

## ✅ **Drop Column with Governed Tags**

You must **remove governed tags first**:

```sql
UNSET TAG ON COLUMN finance_catalog.sales_schema.orders.email pii;

ALTER TABLE finance_catalog.sales_schema.orders
DROP COLUMN email;
```

***

## ✅ **Search Using Tags**

*   Use **workspace search bar** to find tables/views by tag key or value.
*   Example: Search for `sensitivity:PII`.

***

## ✅ **Retrieve Tags via Information Schema**

Query tag metadata:

```sql
SELECT * FROM finance_catalog.INFORMATION_SCHEMA.TABLE_TAGS;
SELECT * FROM finance_catalog.INFORMATION_SCHEMA.COLUMN_TAGS;
```

***

### ✅ **Best Practices**

*   Define a **tagging policy** for consistency.
*   Use **governed tags** for compliance-critical attributes.
*   Combine tags with **comments** for richer metadata.
