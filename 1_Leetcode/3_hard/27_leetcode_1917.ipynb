# 📘 LeetCode 1917: Leetcodify Friends Recommendations

**Level:** Hard  

---

## ❓ Question

We want to recommend friends to Leetcodify users.  

A user `x` should be recommended to user `y` if:  
1. Users `x` and `y` are **not already friends**, and  
2. Users `x` and `y` listened to the same **three or more different songs** on the **same day**.  

Notes:  
- Recommendations are **unidirectional**. If `x` should be recommended to `y`, then `y` should also be recommended to `x`.  
- The result table should not contain duplicates.

````python
%sql
with relation as (
  select user1_id  as self , user2_id as friend from Friendship 

  union 

  select user2_id as self , user1_id  as friend from Friendship 
) , 

cte as(
    Select 
    l1.user_id  as self , l2.user_id  as friend ,
    l1.song_id   ,l1.day          from Listens  l1 inner join Listens  l2 
    on l1.song_id  = l2.song_id 
    and l1.day         = l2.day        
    and l1.user_id  <> l2.user_id 
)  , cte_3 as (
select 
count(*)over(partition by  self, friend , day) as cnt  ,
* from cte 
)
select distinct cte_3.self , cte_3.friend  --,  relation.self    , relation.friend
  from cte_3
left  join relation 
on cte_3.self = relation.self 
and cte_3.friend = relation.friend
where 1=1 
and relation.self  is NULL and relation.friend is NULL
--and self < friend 
and
 cnt >= 3 
````
