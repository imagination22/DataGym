## ✅ **1. CREATE Statements (New Objects)**

    *   `CREATE TABLE`
    *   `CREATE VIEW`

    *   `CREATE CONNECTION`

    *   `CREATE RECIPIENT`

    *   `CREATE SERVER`

## ✅ **3. DROP Statements (Remove Objects)**

*   **Core Structures:**
    *   `DROP CATALOG`
    *   `DROP DATABASE`
    *   `DROP SCHEMA`
    *   `DROP TABLE`
    *   `DROP VIEW`

    *   `DROP PROVIDER`

    *   `DROP SHARE`

***

## ✅ **4. Metadata & Governance**

*   `COMMENT ON` → Add descriptions to objects.
*   `SET TAG` / `UNSET TAG` → Apply or remove tags for classification.

***

## ✅ **5. Maintenance & Refresh**

*   `MSCK REPAIR TABLE` → Sync metadata for partitions.
*   `REFRESH FOREIGN (CATALOG, SCHEMA, TABLE)` → Refresh external metadata.
*   `REFRESH (MATERIALIZED VIEW or STREAMING TABLE)` → Update data.
*   `SYNC` → Synchronize metadata changes.

***

## ✅ **6. Other Operations**

*   `TRUNCATE TABLE` → Remove all rows without dropping structure.
*   `UNDROP TABLE` → Restore a dropped table.

***

### ✅ **Memory Trick**

Think of **CRUD + Governance + Maintenance**:

*   **ALTER** = Change existing
*   **CREATE** = New object
*   **DROP** = Delete object
*   **COMMENT / TAG** = Documentation & classification
*   **REFRESH / REPAIR / SYNC** = Maintenance
*   **TRUNCATE / UNDROP** = Special operations

The message you’re seeing is expected because **Delta Live Tables (DLT)** requires a pipeline to manage and populate streaming tables. Simply running the SQL in a notebook won’t create the table unless it’s part of a DLT pipeline.

***

### ✅ Why This Happens

*   DLT uses **pipelines** to orchestrate ingestion, transformations, and table creation.
*   The `CREATE STREAMING TABLE` statement is valid, but it only works inside a pipeline context.

***

### ✅ How to Fix It

1.  **Create a DLT Pipeline**
    *   In Databricks UI:  
        **Workflows → Delta Live Tables → Create Pipeline**
2.  **Add Your SQL Script**
    *   Use your query:
        ```sql
        CREATE STREAMING TABLE IF NOT EXISTS Student_runs (
          run INT
        );
        ```
3.  **Configure Pipeline Settings**
    *   **Source**: Define the input stream (e.g., Auto Loader, Kafka, or cloud storage).
    *   **Target**: Choose a storage location for the managed tables.
4.  **Start the Pipeline**
    *   Once started, DLT will create and maintain the streaming table.

***

### ✅ Example with Auto Loader

```sql
CREATE STREAMING TABLE Student_runs
COMMENT "Streaming table for student runs"
AS SELECT * FROM cloud_files("/path/to/data", "json");
```
